```python
def calculate_repetition_ratio(text):
    """
    연속 반복 어절 비율을 계산한다.
    반환값은 퍼센트(%) 단위 float.
    """

    tokens = tokenize_text(text)

    if len(tokens) == 0:
        return 0.0

    repeated_count = 0

    for i in range(1, len(tokens)):
        if tokens[i] == tokens[i - 1]:
            repeated_count += 1

    ratio = repeated_count / len(tokens) * 100

    return round(ratio, 2)
```

```python
import re

def tokenize_text(text):
    """
    STT 텍스트를 공백 기준 어절로 분리한다.
    문장부호는 반복어 계산에 방해되므로 제거한다.
    """

    if text is None:
        return []

    text = str(text).strip()

    if text == "":
        return []

    # 문장부호 제거
    text = re.sub(r"[.,!?。！？…~]+", " ", text)

    # 괄호, 따옴표 등 제거
    text = re.sub(r"[\"'“”‘’()\[\]{}<>]", " ", text)

    # 여러 공백 정리
    text = re.sub(r"\s+", " ", text).strip()

    if text == "":
        return []

    return text.split(" ")


def calculate_repetition_ratio(text):
    """
    연속 반복 어절 비율을 계산한다.
    반환값은 퍼센트(%) 단위 float.
    """

    tokens = tokenize_text(text)

    if len(tokens) == 0:
        return 0.0

    repeated_count = 0

    for i in range(1, len(tokens)):
        if tokens[i] == tokens[i - 1]:
            repeated_count += 1

    ratio = repeated_count / len(tokens) * 100

    return round(ratio, 2)


def score_repetition_ratio(repetition_ratio):
    """
    반복어 비율을 0~100점으로 변환한다.
    낮을수록 좋은 점수다.
    """

    if repetition_ratio is None:
        return 0

    if repetition_ratio <= 5:
        return 100
    elif repetition_ratio <= 15:
        return 80
    elif repetition_ratio <= 30:
        return 60
    elif repetition_ratio <= 50:
        return 40
    else:
        return 0


def analyze_repetition(text):
    """
    STT 텍스트에서 반복어 비율과 반복어 점수를 함께 계산한다.
    """

    repetition_ratio = calculate_repetition_ratio(text)
    repetition_score = score_repetition_ratio(repetition_ratio)

    return {
        "repetition_ratio": repetition_ratio,
        "repetition_score": repetition_score
    }
```

```python
samples = [
    "어 어 그 뭐냐 물 주문해브러.",
    "오늘은 유월 유월 사일 같아요.",
    "관리실에 전화해봐야제.",
    "음 음 음 잘 모르겠어요.",
    "둘 다 과일이에요.",
    "",
]

for s in samples:
    result = analyze_repetition(s)

    print("문장:", s)
    print("어절:", tokenize_text(s))
    print("반복어 비율:", result["repetition_ratio"])
    print("반복어 점수:", result["repetition_score"])
    print("-" * 50)
```

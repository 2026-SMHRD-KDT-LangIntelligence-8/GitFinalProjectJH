```python
import json
import pymysql
from pathlib import Path
from getpass import getpass

# =========================
# 1. JSON 파일 설정
# =========================

JSON_FILE = "./questions_split_3_7.json"

json_path = Path(JSON_FILE)

if not json_path.exists():
    raise FileNotFoundError(f"JSON 파일을 찾을 수 없습니다: {JSON_FILE}")

with json_path.open("r", encoding="utf-8") as f:
    questions = json.load(f)

print(f"JSON 문항 수: {len(questions)}")


# =========================
# 2. DB 접속 정보 입력
# =========================

DB_HOST = input("DB host 입력: ").strip()
DB_PORT = int(input("DB port 입력(보통 3306): ").strip() or "3306")
DB_USER = input("DB user 입력: ").strip()
DB_PASSWORD = getpass("DB password 입력: ")
DB_NAME = input("DB name 입력: ").strip()

DB_CONFIG = {
    "host": DB_HOST,
    "port": DB_PORT,
    "user": DB_USER,
    "password": DB_PASSWORD,
    "database": DB_NAME,
    "charset": "utf8mb4"
}


# =========================
# 3. 필수 필드 확인
# =========================

required_fields = [
    "question_type_id",
    "question_text",
    "question_purpose",
    "image_file_path",
    "image_description_criteria"
]

for idx, q in enumerate(questions, start=1):
    missing_fields = [field for field in required_fields if field not in q]

    if missing_fields:
        raise ValueError(
            f"{idx}번째 문항에 필수 필드가 없습니다: {missing_fields}\n문항 내용: {q}"
        )

print("필수 필드 확인 완료")


# =========================
# 4. DB 연결 테스트
# =========================

conn = pymysql.connect(**DB_CONFIG)
cursor = conn.cursor()

print("DB 연결 성공")


# =========================
# 5. 삽입 전 현재 문항 수 확인
# =========================

cursor.execute("SELECT COUNT(*) FROM QUESTIONS")
before_count = cursor.fetchone()[0]

print(f"삽입 전 QUESTIONS 문항 수: {before_count}")

if before_count > 0:
    raise RuntimeError(
        "QUESTIONS 테이블에 이미 데이터가 있습니다. "
        "중복 삽입을 막기 위해 실행을 중단합니다."
    )


# =========================
# 6. INSERT SQL
# question_id는 AUTO_INCREMENT라서 넣지 않음
# =========================

insert_sql = """
INSERT INTO QUESTIONS (
    question_type_id,
    question_text,
    question_purpose,
    image_file_path,
    image_description_criteria
)
VALUES (%s, %s, %s, %s, %s)
"""

insert_data = []

purpose_map = {
    "test": "검사",
    "train": "훈련"
}

for q in questions:
    raw_purpose = q.get("question_purpose")
    db_purpose = purpose_map.get(raw_purpose)

    if db_purpose is None:
        raise ValueError(f"알 수 없는 question_purpose 값: {raw_purpose}")

    insert_data.append((
        q.get("question_type_id"),
        q.get("question_text"),
        db_purpose,
        q.get("image_file_path"),
        q.get("image_description_criteria")
    ))

print(f"INSERT 준비 문항 수: {len(insert_data)}")


# =========================
# 7. DB 삽입
# =========================

try:
    cursor.executemany(insert_sql, insert_data)
    conn.commit()

    print(f"삽입 완료: {cursor.rowcount}개")

except Exception as e:
    conn.rollback()
    print("삽입 중 오류 발생. 변경사항을 취소했습니다.")
    raise e

finally:
    cursor.close()
    conn.close()
    print("DB 연결 종료")
```
